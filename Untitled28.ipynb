```python
# crie um programa que calcule a soma dos números digitados pelo usuário

total_sum = 0
print("Enter numbers to sum. Enter '0' to finish.")

while True:
    try:
        num_str = input("Enter a number: ")
        number = float(num_str)
        if number == 0:
            break
        total_sum += number
    except ValueError:
        print("Invalid input. Please enter a valid number.")

print(f"The sum of the numbers is: {total_sum}")
```

```python
# solicite um número de 1 ate 10 ao usuário e exiba a tabuada

while True:
    try:
        num_str = input("Digite um número de 1 a 10 para ver a tabuada: ")
        number = int(num_str)
        if 1 <= number <= 10:
            break
        else:
            print("Por favor, digite um número entre 1 e 10.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

print(f"Tabuada do {number}:")
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

```python
# solicite ao usuário 4 notas e calcule a media

grades = []
num_grades = 4

for i in range(num_grades):
    while True:
        try:
            grade_str = input(f"Digite a nota {i+1}: ")
            grade = float(grade_str)
            grades.append(grade)
            break
        except ValueError:
            print("Entrada inválida. Por favor, digite um número para a nota.")

if grades:
    average = sum(grades) / len(grades)
    print(f"As notas digitadas foram: {grades}")
    print(f"A média das notas é: {average:.2f}")
else:
    print("Nenhuma nota foi digitada.")
```

```python
# faça um programa que receba valores e quantidades de produtos e no final exiba o valor total da compra




total_purchase_value = 0.0

print("Calculadora de valor total da compra. Digite 'fim' quando terminar.")

while True:
    price_input = input("Digite o preço do produto (ou 'fim' para finalizar): ").lower()
    if price_input == 'fim':
        break

    try:
        price = float(price_input)
        if price < 0:
            print("O preço não pode ser negativo. Tente novamente.")
            continue

        quantity_input = input("Digite a quantidade do produto: ")
        quantity = int(quantity_input)
        if quantity < 0:
            print("A quantidade não pode ser negativa. Tente novamente.")
            continue

        item_total = price * quantity
        total_purchase_value += item_total
        print(f"Produto adicionado: {quantity} x R${price:.2f} = R${item_total:.2f}")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número para o preço e um número inteiro para a quantidade.")
    except Exception as e:
        print(f"Ocorreu um erro: {e}")

print(f"\nValor total da compra: R${total_purchase_value:.2f}")
```

```python
# peça para o usuário digitar 5 números e no final exibir uma mensagem com o maior número



numbers = []
num_count = 5

print(f"Por favor, digite {num_count} números.")

for i in range(num_count):
    while True:
        try:
            num_str = input(f"Digite o número {i+1}: ")
            number = float(num_str)
            numbers.append(number)
            break
        except ValueError:
            print("Entrada inválida. Por favor, digite um número válido.")

if numbers:
    largest_number = max(numbers)
    print(f"Os números digitados foram: {numbers}")
    print(f"O maior número digitado é: {largest_number}")
else:
    print("Nenhum número foi digitado.")
```

```python
import random

secret_number = random.randint(1, 10)
print("Eu escolhi um número entre 1 e 10. Tente adivinhar!")
```

```python
# crie um jogo de adivinhação, o usuário tenta adivinhar um número secreto de 1 ate 10.



guess = 0
while guess != secret_number:
    try:
        guess_str = input("Digite seu palpite (um número entre 1 e 10): ")
        guess = int(guess_str)

        if 1 <= guess <= 10:
            if guess < secret_number:
                print("Seu palpite é muito baixo. Tente novamente!")
            elif guess > secret_number:
                print("Seu palpite é muito alto. Tente novamente!")
            else:
                print(f"Parabéns! Você acertou o número secreto, que era {secret_number}!")
        else:
            print("Por favor, digite um número entre 1 e 10.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")
```
