```python
# escrever um programa que o usuário digita um número e o programa inprime de 0 até o número

# entrada

número = int(input("digite um número"))

# processamento/saida

for i in range (número):
  print(i)
```

```python
# escrever um programa que o usuário digita um número e o programa inprime em ordem decrescente
# até o zero

# entrada

número = int(input("digite um número"))

# processamento/saida

for i in range (número, -1, -1):
  print(i)
```

```python

# peça um número e mostre todos os números pares até o número

# entrada

número = int(input("digite um número"))

# processamento/saida

for i in range(número + 1):
  if i % 2 == 0:
    print(i)
```

```python
# Peça um número e calcule a soma de 1 até o número

# entrada

número = int(input("digite um número"))

# processamento/saida

soma = 0
for i in range(1, número + 1):
  soma += i

print(f"A soma de 1 até {número} é: {soma}")
```

```python
# peça uma palavrav e conte as letras: EX.; Phyton - A palavra tem 06 letras

# entrada
palavra = input("Digite uma palavra: ")

# processamento/saida
num_letras = len(palavra)
print(f"A palavra '{palavra}' tem {num_letras} letras")
```

```python
# peça uma palavra e conte as vogais

# entrada
palavra = input("Digite uma palavra: ")

# processamento/saida
vogais = "aeiouAEIOU"
contador_vogais = 0

for letra in palavra:
  if letra in vogais:
    contador_vogais += 1

print(f"A palavra '{palavra}' tem {contador_vogais} vogais")
```

```python
# Peça ao usuário para digitar 05 números e informe o maior

# entrada

maior_numero = float('-inf') # Inicializa com o menor número possível

for i in range(5):
  while True:
    try:
      numero = float(input(f"Digite o {i+1}º número: "))
      break
    except ValueError:
      print("Entrada inválida. Por favor, digite um número.")

  # processamento
  if numero > maior_numero:
    maior_numero = numero

# saida
print(f"O maior número digitado foi: {maior_numero}")
```
