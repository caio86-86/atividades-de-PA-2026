```python
# calculadora simples - Escrever um programa que calcula as 04 operações aritméticas

print("......MENU.....")
print("")
print("opção 1 - somar")
print("opção 2 - subtrair")
print("opção 3 - multiplicar")
print("opção 4 - dividir")
print("opção 5 - sair")
print("....................")
print("")

opção = 0

try:
  while opção != 5:
    opção = int(input("digite a opção desejada ou 5 para sair: "))

    if opção == 1:
      a = float(input("digite o primeiro número: "))
      b = float(input("digite o segundo número: "))
      print(f"total da soma de {a} + {b} é igual: {a + b}")

    elif opção == 2:
      a = float(input("digite o primeiro número: "))
      b = float(input("digite o segundo número: "))
      print(f"total da subtração de {a} - {b} = {a - b}")

    elif opção == 3:
      a = float(input("digite o primeiro número: "))
      b = float(input("digite o segundo número: "))
      print(f"O resultado da multiplicação de {a} * {b} = {a * b}")

    elif opção == 4:
      a = float(input("digite o primeiro número: "))
      b = float(input("digite o segundo número: "))
      if b != 0:
        print(f"O total da divisão de {a} / {b} = {a / b}")
      else:
        print("Erro: Divisão por zero!")

    print("")
except ValueError:
  print("Erro: Por favor, digite apenas números.")
```

```python
# escrever um programa que recebe 4 notas, calcule a média e imprime se o aluno foi aprovado, recuperação ou reprovado.

# nota maior ou igual a 7 = aprovado
# nota maior ou igual a 5 = recuperação
# nota menor que 5 = reprovado


nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))
nota4 = float(input("Digite a quarta nota: "))

media = (nota1 + nota2 + nota3 + nota4) / 4

print(f"A média do aluno é: {media:.2f}")

if media >= 7:
  print("Aluno APROVADO!")
elif media >= 5:
  print("Aluno em RECUPERAÇÃO!")
else:
  print("Aluno REPROVADO!")

















```
