```python
# escrever um programa que o usuário digite um número de 1 a 10 e o programa imprime a tabuada do número digitado

try:
    num = int(input("Digite um número de 1 a 10 para ver a tabuada: "))

    if 1 <= num <= 10:
        print(f"Tabuada do {num}:")
        for i in range(1, 11):
            print(f"{num} x {i} = {num * i}")
    else:
        print("Número inválido. Por favor, digite um número entre 1 e 10.")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# escrever um programa que o usuário digita um número e o programa imprime até o número digtiado

try:
    num = int(input("Digite um número inteiro: "))

    if num >= 1:
        print(f"Imprimindo números de 1 até {num}:")
        for i in range(1, num + 1):
            print(i)
    else:
        print("Por favor, digite um número inteiro positivo.")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# escrever um programa que o usário digite um número e o programa imprime o número digitado até zero

try:
    num = float(input("Digite um número: "))

    print(f"Contando de {num} até 0:")
    if num >= 0:
        while num >= 0:
            print(int(num))
            num -= 1
    else:
        while num <= 0:
            print(int(num))
            num += 1
except ValueError:
    print("Entrada inválida. Por favor, digite um número.")
```

```python
# escrever um programa que o usuário digita um número e o programa imprime os núemros pares

try:
    num = int(input("Digite um número: "))

    print(f"Números pares de 0 até {num}:")
    for i in range(0, num + 1):
        if i % 2 == 0:
            print(i)
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# escrever um programa que o usário digita um número e o programa some os números até o número digitado

try:
    num = int(input("Digite um número: "))

    if num >= 1:
        soma = sum(range(1, num + 1))
        print(f"A soma de todos os números de 1 até {num} é: {soma}")
    else:
        print("Por favor, digite um número inteiro positivo.")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# escrever um programa que o usuário digita 04 números e o programa imprime o maior

try:
    num1 = float(input("Digite o primeiro número: "))
    num2 = float(input("Digite o segundo número: "))
    num3 = float(input("Digite o terceiro número: "))
    num4 = float(input("Digite o quarto número: "))

    maior_numero = max(num1, num2, num3, num4)

    print(f"O maior número digitado é: {maior_numero}")
except ValueError:
    print("Entrada inválida. Por favor, digite números válidos.")
```

```python
# Piramide matemática simetrica

# faça um programa que receba um núemro N e imprima uma piramide perfeitamente centralizada

# exemplo para N = 5

#                1
#               212
#              32123
#             4321234
#            543212345

try:
    N = int(input("Digite um número inteiro para a altura da pirâmide (N): "))

    if N < 1:
        print("Por favor, digite um número inteiro positivo para a altura da pirâmide.")
    else:
        for i in range(1, N + 1):
            # Calculate leading spaces for centering
            leading_spaces = " " * (N - i)

            # Construct the left side (counting down)
            left_side = ""
            for j in range(i, 0, -1):
                left_side += str(j)

            # Construct the right side (counting up, starting from 2)
            right_side = ""
            for j in range(2, i + 1):
                right_side += str(j)

            # Combine and print the row
            print(f"{leading_spaces}{left_side}{right_side}")

except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```
