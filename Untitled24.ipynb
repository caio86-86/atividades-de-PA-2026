```python
# escrever um programa que o usuário digita um número de 1 a 10 e o programa imprime a tabuada do número digitado.

```

```python
try:
    num = int(input("Digite um número de 1 a 10 para ver a tabuada: "))

    if 1 <= num <= 10:
        print(f"Tabuada do {num}:")
        for i in range(1, 11):
            print(f"{num} x {i} = {num * i}")
    else:
        print("Número inválido. Por favor, digite um número entre 1 e 10.")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# escrever um programa que o usuário digita um número e o programa imprime até o número digitado

```

```python
try:
    num = int(input("Digite um número inteiro: "))

    if num >= 1:
        print(f"Imprimindo números de 1 até {num}:")
        for i in range(1, num + 1):
            print(i)
    else:
        print("Por favor, digite um número inteiro positivo.")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```
